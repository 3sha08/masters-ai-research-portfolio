# **Research Findings**

This study compared the performance of three leading Large Language Models—ChatGPT (GPT-5.5), Google Gemini 3.6 Flash, and Claude Sonnet—using an identical business summary generated from the Tableau Sample Superstore dataset and a standardized evaluation prompt.

All three models demonstrated strong business understanding, numerical accuracy, and evidence-based reasoning by relying primarily on the provided business summary without introducing unsupported assumptions.

ChatGPT produced a well-structured executive report with clear recommendations and appropriately identified data limitations. Its responses were concise, highly readable, and maintained strict adherence to the supplied information.

Gemini distinguished itself through detailed quantitative analysis, calculating additional business metrics such as profit margin, revenue contribution percentages, and regional profit shares. These derived calculations provided additional business insights while remaining consistent with the supplied data.

Claude generated the most comprehensive executive analysis by combining detailed numerical reasoning with highly structured business recommendations. It provided extensive derived metrics, explicit validation of reported totals, and carefully documented data limitations, making its response particularly suitable for executive decision support.

Overall, all three models produced high-quality outputs. Claude demonstrated the strongest analytical depth, Gemini excelled in quantitative business analytics, and ChatGPT provided the most balanced combination of readability, evidence-based reasoning, and executive communication.
